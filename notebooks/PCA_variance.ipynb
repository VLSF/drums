```python
from pathlib import Path
import numpy as np
import pandas as pd

root = Path.cwd()
if not (root / "triangle_drums.npz").exists():
    root = root.parent
spectra = np.load(root / "triangle_drums.npz")["spectra"]
logs = np.log(spectra[:int(0.9 * len(spectra))].astype(np.float32))
X = (logs - logs.mean(axis=0)) / logs.std(axis=0)
X -= X.mean(axis=0)

_, singular_values, Vt = np.linalg.svd(X, full_matrices=False)
variance_share = singular_values**2 / np.sum(singular_values**2)
print(f"X: {X.shape}; V: {Vt.T.shape}")
print(f"Первая компонента: {100 * variance_share[0]:.4f}% дисперсии")
print(f"Остальные 29 вместе: {100 * variance_share[1:].sum():.4f}%")
```
